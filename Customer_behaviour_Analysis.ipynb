# feature engineering

### in this data, we want to add columns that grounps data in some categories like:
young adult , adult ,  middle_aged ,  senior


```python
labels=['young adult' , 'adult' ,  'middle_aged' ,  'senior']
df['age_groups']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_groups','review_rating']].head(4)
```

## we also creating new column for frequently purchasing_days  , even the coiulmn is available we are replacing text with numbers 

```python
df['frequency_of_purchases'].unique()
```

```python
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Annually' : 365,
    'Quarterly' : 90,
    'Bi-Weekly': 14,
       'Monthly' : 30, 
    'Every 3 Months': 90
    
}

df['purchase_frequently_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
(df[['purchase_frequently_days','frequency_of_purchases']]).head(5)
```
